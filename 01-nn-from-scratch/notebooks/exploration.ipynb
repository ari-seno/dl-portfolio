```python
import numpy as np
import matplotlib.pyplot as plt
from src.train import load_data, train
from src.model import NeuralNetwork

X, y = load_data(n_samples=500, noise=0.15, seed=42)
```

```python
model = NeuralNetwork(input_dim=2, hidden_dim=32, output_dim=1)
losses = train(model, X, y, epochs=4000, learning_rate=0.3, print_every=500)
```

```python
plt.figure(figsize=(6, 4))
plt.plot(losses)
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Loss Curve")
plt.savefig("outputs/loss_curve.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
def plot_decision_boundary(model, X, y):
    x_min, x_max = X[:, 0].min() - 0.5, X[:, 0].max() + 0.5
    y_min, y_max = X[:, 1].min() - 0.5, X[:, 1].max() + 0.5
    xx, yy = np.meshgrid(np.linspace(x_min, x_max, 300),
                          np.linspace(y_min, y_max, 300))

    grid = np.c_[xx.ravel(), yy.ravel()]
    Z = model.predict(grid).reshape(xx.shape)

    plt.figure(figsize=(6, 5))
    plt.contourf(xx, yy, Z, alpha=0.3, cmap="coolwarm")
    plt.scatter(X[:, 0], X[:, 1], c=y.ravel(), cmap="coolwarm", edgecolors="k")
    plt.xlabel("Feature 1")
    plt.ylabel("Feature 2")
    plt.title("Decision Boundary — Neural Network from Scratch (99.8% acc)")
    plt.savefig("outputs/decision_boundary.png", dpi=150, bbox_inches="tight")
    plt.show()

plot_decision_boundary(model, X, y)
```
